```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv")
df.head()
df.describe()
```

```python
import pandas as pd
from scipy import stats

beans = df[df["foods"] == "Beans"]["glucose"]
rice = df[df["foods"] == "Rice"]["glucose"]

t_stat, p_value = stats.ttest_ind(beans, rice)

print("P-value:", p_value)
```

```python
import pandas as pd
from scipy import stats

# Load datasets
cgm_df = pd.read_csv("/Users/nikhilkumargarg/Downloads/data_cgm.csv")
smbg_df = pd.read_csv("/Downloads/data_smbg.csv")

# Select glucose columns
cgm = cgm_df["glucose"]
smbg = smbg_df["glucose"]

# Hypothesis test
t_stat, p_value = stats.ttest_rel(cgm, smbg)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
from scipy.stats import pearsonr

corr, p_value = pearsonr(df["mins_since_start"], df["glucose"])

print("Correlation:", corr)
print("P-value:", p_value)
```

```python
beans = df[df["foods"] == "Beans"]["glucose"]
rice = df[df["foods"] == "Rice"]["glucose"]
bread = df[df["foods"] == "Bread"]["glucose"]

from scipy.stats import f_oneway

f_stat, p_value = f_oneway(beans, rice, bread)

print("P-value:", p_value)
```

```python
from scipy import stats

t_stat, p_value = stats.ttest_1samp(df["glucose"], 100)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```
